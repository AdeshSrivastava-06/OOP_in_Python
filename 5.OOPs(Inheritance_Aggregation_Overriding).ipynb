Aggregation(Has-A relationship)

```python
class customer:
    
    def __init__(self,name,gender,address):
        self.name = name
        self.gender = gender
        self.address = address
        
    def print_address(self):
        
        print(self.address.get_city(),self.address.state,self.address.country)
        
    def edit_profile(self,new_name,new_city,new_state,new_country):
        self.name = new_name
        self.address.edit_address(new_city,new_state,new_country)
         
class Address:
    
    def __init__(self,city,state,country):
        self.__city = city
        self.state = state
        self.country = country
        
    def get_city(self):
        return self.__city
    
    def edit_address(self,new_city,new_state,new_country):
        self.__city = new_city
        self.state = new_state
        self.country = new_country
        
add1 = Address("Noida","UP","India")
        
cust1 = customer("Adesh","Male",add1)
cust1.print_address()

#this is an example of aggregation where customer class has an instance of Address class.
#While performing aggregation, we cannot access the private attributes of other class, so we can use getter and setter methods. 
```

```python
class User:
    
    def __init__(self,name):
        self.name = name
        
    def login(self):
        print(f"{self.name} logged in")

class Student(User):
    
    def __init__(self,roll_no,name):
        super().__init__(name) #super() is used to call the constructor of the parent class.
        #If our child class does not have any constructor, then by default the parent's constructor is called. But if our child class has a constructor in it, then we need to call the parent class constructor by using the `super` keyword. Otherwise, only the child class constructor will be there, and no parent class. 
        self.roll_no = roll_no
        
    def display(self):
        print(f"Name: {self.name}, Roll No: {self.roll_no}") 
        
User1 = User("Adesh")

student1 = Student(101, "Adesh")

student1.login()
student1.display()
```

```python
class Phone:
    
    def __init__(self,price,brand,camera):
        self.price = price
        self.brand = brand
        self.camera = camera
        print("Inside Phone class constructor")
        
    def buy(self):
        print(f"Buying {self.brand} phone for {self.price} of camera {self.camera}")
        
class SmartPhone(Phone):
    
    pass 

s = SmartPhone(20000,"Samsung","12MP")
s.buy() #This is an example of inheritance where SmartPhone class inherits the properties and methods of Phone class. Since SmartPhone class does not have any constructor, the constructor of Phone class is called.
s.camera = "16MP" #We can also change the value of the attributes of the parent class in the child class.
print(s.camera)
```

```python
"""A child class cannot directly access a parent's private variable.

But a child class can use any parent method (getter, setter, or any other method) that accesses that private variable, because the access is happening inside the parent class's code."""

class Phone:
    
    def __init__(self,price,brand,camera):
        self.__price = price
        self._brand = brand
        self._camera = camera
        print("Inside Phone class constructor")
        
    def buy(self):
        
        print(self.__price,self._brand,self._camera)
        
class SmartPhone(Phone):
    
    def check_price(self):
        print(self.__price) 
        
s = SmartPhone(20000,"Samsung","12MP")
s.buy()#This will work because the buy method is defined in the parent class and it can access the private variable __price.
s.check_price() #This will give an error because __price is a private variable and cannot be accessed directly in the child class. We can use a getter method to access it.
```

```python
class A:
    
    def __init__(self):
        self.var1 = 100

    def display1(self, var1):
        print("class A :", self.var1)


class B(A):

    def display2(self, var1):
        print("class B :", self.var1)


obj = B()

obj.display1(200)

#self.var1 refers to the object's instance variable initialized by the constructor (100), while var1 is just a local method parameter (200); since the code prints self.var1, the output is 100.
```

```python
class A:
    
    def __init__(self):
        self.var1 = 100

    def display1(self, var1):
        self.var1 = var1
        print("class A :", self.var1)


class B(A):

    def display2(self, var1):
        print("class B :", self.var1)


obj = B()

obj.display1(200)

#here the output will be 200 because we are assigning the value of var1 to self.var1 in the display1 method of class A. So, when we call the display2 method of class B, it will print the value of self.var1 which is now 200.
```

Super Keyword

```python
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class SmartPhone(Phone):
    def buy(self):
        super().buy() #This will call the buy method of the parent class and then we can add our own code in the child class's buy method.
        print ("Buying a smartphone")

s=SmartPhone(20000, "Apple", 13)

s.buy()
```

```python
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.price = price
        self.brand = brand
        self.camera = camera

class SmartPhone(Phone):
    def __init__(self, price, brand, camera, os, ram):
        print('Inside smartphone constructor')
        super().__init__(price, brand, camera)
        self.os = os
        self.ram = ram
        print ("Inside smartphone constructor")

s=SmartPhone(20000, "Samsung", 12, "Android", 2)

print(s.os)
print(s.brand)
print(s.camera)

#Super is used inside the class and not outside the class, generally in the child class. 
#A super keyword is used to access the methods and not the attributes or the variables. like we cannot use super().price to access the price variable of the parent class, but we can use super().buy() to access the buy method of the parent class.
```

```python
class Parent:

    def __init__(self,num):
      self.__num=num

    def get_num(self):
      return self.__num

class Child(Parent):
  
    def __init__(self,num,val):
      super().__init__(num)
      self.__val=val

    def get_val(self):
      return self.__val
      
son=Child(100,200)
print(son.get_num())
print(son.get_val())
```

```python
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product:
    def buy(self):
        print ("Product buy method")

# Method resolution order it is the order in which the methods are resolved in case of multiple inheritance. In this case, the buy method of Phone class will be called because it is the first class in the inheritance list of SmartPhone class.
class SmartPhone(Phone,Product):
    pass

s=SmartPhone(87700, "Samsung", 17)

s.buy()
```

```python
class A:

    def m1(self):
        return 20

class B(A):

    def m1(self):
        val=super().m1()+30
        return val

class C(B):
  
    def m1(self):
        val=super().m1()+20 #self.m1() will call the m1 method of class C which will lead to infinite recursion. So, we need to use super() to call the m1 method of class B.
        return val
obj=C()
print(obj.m1())

#it will give an error because the m1 method of class C is calling itself recursively without any base case, which will lead to infinite recursion. To fix this, we can call the m1 method of class B using super() instead of calling self.m1() in class C.

#the super() function is used to call the method of the parent class. In this case, we are calling the m1 method of class B from class C using super(). This way, we can avoid infinite recursion and get the correct output.
```

Polymorphism

1.Method overriding
2.method overloading
3.operator overloading

```python
class Shape:
    
    def area(self,radius):
        return 3.14*radius*radius
    
    def area(self,length,breadth):
        return length*breadth

#it makes the code clean 

circle = Shape()

print(circle.area(5,6)) 
print(circle.area(5)) #This will give an error because the area method is defined twice in the Shape class, and the second definition will override the first one. So, when we call the area method with one argument, it will try to call the second definition of the area method which expects two arguments, and it will give an error. To fix this, we can use method overloading by defining the area method with different number of parameters. However, Python does not support method overloading directly, but we can achieve it by using default parameters or by checking the number of arguments passed to the method.
```

```python
class Shape:
    
    def area(self,length,breadth=0):
        if breadth == 0:
            return 3.14 * length * length
        else:
            return length * breadth
        
circle = Shape()
print(circle.area(5)) 
print(circle.area(5,6)) 
```

```python
#operator overloading

class Point:
    
    def __init__(self,x,y):
        self.x = x
        self.y = y
        
    def __add__(self,other):
        return Point(self.x + other.x, self.y + other.y)
    
p1 = Point(2,3)
p2 = Point(4,5)
p3 = p1 + p2
print(f'Addition of two points: ({p3.x}, {p3.y})')
print(p3.x, p3.y)
```

Abstraction

```python
#Abstraction is the process of hiding the implementation details and showing only the functionality to the user. It is achieved by using abstract classes and abstract methods. An abstract class is a class that cannot be instantiated and is meant to be inherited by other classes. An abstract method is a method that is declared in the abstract class but does not have any implementation. The child class that inherits the abstract class must implement the abstract method.

#it basically enforces the child class to implement the abstract method of the parent class then only the child class can be instantiated.It can be written for security reasons or to hide the implementation details from the user. It is also used to achieve polymorphism. In Python, we can use the abc module to create abstract classes and abstract methods.

#we cannot make a object of an abstract class


from abc import ABC, abstractmethod

class BankApp(ABC):
    
    def database_connection(self):
        print("Connecting to database")
        
    @abstractmethod
    def transfer_money(self):
        pass

class Paytm(BankApp):
    
    def login(self):
        print("Logging in to Paytm")
        
```

```python
#the above code will give an error because the Paytm class is not implementing the abstract method transfer_money() of the BankApp class. To fix this, we need to implement the transfer_money() method in the Paytm class.

from abc import ABC, abstractmethod

class BankApp(ABC):
    
    def database_connection(self):
        print("Connecting to database")
        
    @abstractmethod
    def transfer_money(self):
        pass

class Paytm(BankApp):
    
    def login(self):
        print("Logging in to Paytm")
        
    def transfer_money(self):
        print("Transferring money using Paytm")
        
p1 = Paytm()
p1.database_connection()
p1.login()
p1.transfer_money()


```
