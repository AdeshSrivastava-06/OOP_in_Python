```python
#Write OOP classes to handle the following scenarios:

# A user can create and view 2D coordinates
# A user can find out the distance between 2 coordinates
# A user can find the distance of a coordinate from origin
# A user can check if a point lies on a given line
# A user can find the distance between a given 2D point and a given line
```

```python
class point:
    
    def __init__(self,x,y):
        self.x_cod = x
        self.y_cod = y
        
    def __str__(self):
        return '<{},{}>'.format(self.x_cod,self.y_cod)
    
    def distance_dis(self,other):
        
        return ((self.x_cod - other.x_cod)**2 + (self.y_cod - other.y_cod)**2)**0.5
    
    def distance_origin(self):
        return (self.x_cod**2 + self.y_cod**2)**0.5
        #other method 
        #return self.distance_dis(point(0,0))
        
class line:
    
    def __init__(self,point1,point2,point3):
        self.p1 = point1
        self.p2 = point2
        self.p3 = point3
    
    def __str__(self):
        
        return '{}x + {}y + {} = 0'.format(self.p1,self.p2,self.p3)
    
    def point_on_line(self,point):
        
        if (self.p1*point.x_cod + self.p2*point.y_cod + self.p3) == 0:
            return "Point lies on the line"
        else:
            return "Point does not lie on the line"

    def distance_point_line(self,point):
        
        return abs(self.p1*point.x_cod + self.p2*point.y_cod + self.p3)/((self.p1**2 + self.p2**2)**0.5)
    
    def does_line_intersect(self,other):
        
        determinant = self.p1*other.p2 - other.p1*self.p2
        
        if determinant == 0:
            return "Lines are parallel"
        else:
            return "Lines intersect"
    
    def intersection_line(self,other):
        
        determinant = self.p1*other.p2 - other.p1*self.p2
        
        if determinant == 0:
            return "Lines are parallel"
        else:
            x = (other.p2*(-self.p3) - self.p2*(-other.p3))/determinant
            y = (self.p1*(-other.p3) - other.p1*(-self.p3))/determinant
            return point(x,y)
    
```

```python
l1 = line(1,1,-2)
l2 = line(2,3,-5)
p1 = point(1,2)
print(l1)
print(p1)
print(l1.point_on_line(p1))
print(l1.distance_point_line(p1))
print(f'Lines l1 and l2 {l1.does_line_intersect(l2)}')
print(f'Intersection of {l1} and {l2} is: {l1.intersection_line(l2)}')
```
