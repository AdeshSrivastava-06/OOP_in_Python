```python
#small banking application
#The class name is written in PascalCase. The first letter of each word is capitalized.
class Atm:
    
    #constructor(A constructor is a special piece of code that we don't have to call explicitly to execute like other functions. it is automatically called when we create an object of a class. It is used to initialize the attributes of a class.)
    def __init__(self):
        
        self.pin = ""
        self.balance = 0
        self.menu()
        
    def menu(self):
        
        while True:
            user_input = input("""
                Hi there! Welcome to the ATM machine.
                
                1. press 1 to create a new pin.
                2. Press 2 to change pin.
                3. Press 3 to check balance.
                4. Press 4 to withdraw money. 
                5. Anything else to exit.
                """)
            
            if user_input == "1":
                #create pin
                self.create_pin()
            elif user_input == "2":
                #change pin
                self.change_pin()
                pass
            elif user_input == "3":
                #check balance
                self.check_balance()
                pass
            elif user_input == "4":
                #withdraw money
                self.withdraw_money()
                pass
            else:
                print("Thank you for using the ATM machine. Have a nice day!")
                break
            
    def create_pin(self):
        
        input_pin = input("Enter your new pin: ")
        self.pin = input_pin
        
        input_balance = int(input("Enter your initial balance: "))
        self.balance = input_balance
        
        print("Your pin has been created successfully!")

        
    def change_pin(self):
        
        input_pin = input("Enter your current pin: ")
        
        if input_pin == self.pin:
            new_pin = input("Enter your new pin: ")
            self.pin = new_pin
            print("Your pin has been changed successfully!")
        else:
            print("Incorrect pin. Please try again.")
            
    def check_balance(self):
        
        input_pin = input("Enter your pin: ")
        
        if input_pin == self.pin:
            print(f"Your balance is: {self.balance}")
        else:
            print("Incorrect pin. Please try again.")
            
    def withdraw_money(self):
        
        input_pin = input("Enter your pin: ")
        
        if input_pin == self.pin:
            amount = int(input("Enter the amount to withdraw: "))
            
            if amount <= self.balance:
                self.balance -= amount
                print(f"You have withdrawn {amount}. Your new balance is: {self.balance}")
            else:
                print("Insufficient balance.")
        else:
            print("Incorrect pin. Please try again.")
        
        
```

```python
obj = Atm()

print(type(obj))  # <class '__main__.Atm'>
```

```python
#Creating our own data type fraction 

class Fraction:
    
    #parameterized constructor
    
    def __init__(self,x,y):
        self.numerator = x
        self.denominator = y
        
    #Whenever I want to print my object, so it will, by default, go to this magic method. this method is used to represent the object in a string format.
        
    def __str__(self):
        return f"{self.numerator}/{self.denominator}"
    
    #add is a magic method which is used to add two objects of the same class.here we can write our custom logic to add two objects of the same class.
    
    #Add requires 2 objects of the same class. So, we have to pass 2 objects of the same class as parameters. The first object is passed automatically as self and the second object is passed as other.
    
    #We can use anything apart from self and other. because these are just parameters. But, by convention, we use self and other.
    def __add__(self,other):
        new_numerator = self.numerator * other.denominator + self.denominator * other.numerator
        new_denominator = self.denominator * other.denominator
        
        return Fraction(new_numerator,new_denominator)
    
    def __sub__(self,other):
        new_numerator = self.numerator * other.denominator - self.denominator * other.numerator
        new_denominator = self.denominator * other.denominator
        
        return Fraction(new_numerator,new_denominator)
    
    def __mul__(self,other):
        new_numerator = self.numerator * other.numerator
        new_denominator = self.denominator * other.denominator
        
        return Fraction(new_numerator,new_denominator)
    
    def __truediv__(self,other):
        new_numerator = self.numerator * other.denominator
        new_denominator = self.denominator * other.numerator
        
        return Fraction(new_numerator,new_denominator)
    
    #custom magic method
    
    def convert_to_decimal(self):
        return self.numerator / self.denominator
```

```python
f1 = Fraction(3,4)
f2 = Fraction(5,6)
print(f'f1 + f2 = {f1 + f2}') 
print(f'f1 - f2 = {f1 - f2}') 
print(f'f1 * f2 = {f1 * f2}') 
print(f'f1 / f2 = {f1 / f2}')  

print(f'f1 in decimal = {f1.convert_to_decimal()}')
#ADD to decimal
print(f'Add in decimal = {f1.convert_to_decimal() + f2.convert_to_decimal()}')
```
