```python
"""Static variables VS instance variables"""

"""Instance variables are unique to each instance of a class. Each object of the class has its own copy of the instance variable.It's a variable of object"""
"""Static variables are shared among all instances of a class. There is only one copy of the static variable, and it is shared by all objects of the class.Its a variable of class Static variable is the same for every object. """

"""classname.variable name to access static variable. objectname.variable name to access instance variable. """

#small banking application
#The class name is written in PascalCase. The first letter of each word is capitalized.
class Atm:
    
    __counter = 1 #static variable
    
    #constructor(A constructor is a special piece of code that we don't have to call explicitly to execute like other functions. it is automatically called when we create an object of a class. It is used to initialize the attributes of a class.)
    def __init__(self):
        
        self.__pin = "" #private variable
        self.__balance = 0
        self.cid = Atm.__counter
        Atm.__counter += 1
        #self.menu()
    
    #utility functions to access static variable
    #here we dont have to create an object to access the static variable, we can directly access it by using the class name.
    @staticmethod 
    def get_id():
        return Atm.__counter
    
    @staticmethod
    def set_id(new_id):
        Atm.__counter = new_id
           
    def get_balance(self):
        return self.__balance
        
    def set_balance(self, new_balance):
        
        if type(new_balance) == int:
            self.__balance = new_balance
        else:
            print("Please enter a valid integer value for balance.")
            
    def menu(self):
        
        while True:
            user_input = input("""
                Hi there! Welcome to the ATM machine.
                
                1. press 1 to create a new pin.
                2. Press 2 to change pin.
                3. Press 3 to check balance.
                4. Press 4 to withdraw money. 
                5. Anything else to exit.
                """)
            
            if user_input == "1":
                #create pin
                self.create_pin()
            elif user_input == "2":
                #change pin
                self.change_pin()
                pass
            elif user_input == "3":
                #check balance
                self.check_balance()
                pass
            elif user_input == "4":
                #withdraw money
                self.withdraw_money()
                pass
            else:
                print("Thank you for using the ATM machine. Have a nice day!")
                break
            
    def create_pin(self):
        
        input_pin = input("Enter your new pin: ")
        self.__pin = input_pin
        
        input_balance = int(input("Enter your initial balance: "))
        self.__balance = input_balance
        
        print("Your pin has been created successfully!")

        
    def change_pin(self):
        
        input_pin = input("Enter your current pin: ")
        
        if input_pin == self.__pin:
            new_pin = input("Enter your new pin: ")
            self.__pin = new_pin
            print("Your pin has been changed successfully!")
        else:
            print("Incorrect pin. Please try again.")
            
    def check_balance(self):
        
        input_pin = input("Enter your pin: ")
        
        if input_pin == self.__pin:
            print(f"Your balance is: {self.__balance}")
        else:
            print("Incorrect pin. Please try again.")
            
    def withdraw_money(self):
        
        input_pin = input("Enter your pin: ")
        
        if input_pin == self.__pin:
            amount = int(input("Enter the amount to withdraw: "))
            
            if amount <= self.__balance:
                self.__balance -= amount
                print(f"You have withdrawn {amount}. Your new balance is: {self.__balance}")
            else:
                print("Insufficient balance.")
        else:
            print("Incorrect pin. Please try again.")
        
        
```
