```python
class Person:
    
    def __init__(self,name,country):
        self.name = name
        self.country = country
        
    def greet(self):
        
        if self.country == 'India':
            print(f'Namaste {self.name}')
            
        else:
            print(f'Hello {self.name}')
```

```python
#if we create an object of class then we can aslo access the attributes of class using that object

person1 = Person('Adesh','India')
print(person1.greet()) # Namaste Adesh
print(person1.name) # Adesh
print(person1.country) # India
```

```python
#We can not access the attributes or methods that are not written in the class. but

person1.age = 25 # we can add new attributes to the object but it will not be available for other objects of the class
print(person1.age) # 25
```

```python
#From the above, we can see that we have created an object without a reference. Whenever we are writing an object name like p1 is equal to person, here scientifically p1 is not the object, but it is a variable that stores the address of the object created by calling that class. so, when we are writing person('Adesh', 'India'), here we are creating an object but we are not storing the address of that object in any variable. so, that object is created but we can not access it because we do not have the reference of that object. so, it is not a good practice to create an object without a reference.
```

```python
"""Pass by reference and pass by value"""

#In Python, when we pass an object to a function, we are passing a reference to that object. This means that if we modify the object inside the function, it will affect the original object outside the function. This is known as "pass by reference".

class Person:
    
    def __init__(self,name,gender):
        self.name = name
        self.gender = gender
        
#outside the class-function that takes an object as a parameter

def greet(person):
    
    print("Hi my name is", person.name, "and I am", person.gender)
    p1 = Person('Adesh', 'Male')
    return p1 
    
person1 = Person('Adesh', 'Male')
x = greet(person1)
print(x.name) 
#in python we can pass our object as a parameter to a function and we can also return an object from a function

#This is called pass by reference because when we are passing a parameter or object as a parameter to a function, then the object's reference ID is passed. The ID of the object and the main class will be the same, so it is a pass by reference. 
```

```python
class Person:
    
    def __init__(self,name,gender):
        self.name = name
        self.gender = gender
        
#outside the class-function that takes an object as a parameter

def greet(person):
    
    person.name = 'Adesh Srivastava'
    
p2 = Person('Aditya', 'Male')
greet(p2)
print(p2.name)  
#why - Because Python passes objects by reference, so modifying the object inside the function affects the original object outside the function.
```

```python
#Mutability of objects in Python

class Person:
    
    def __init__(self,name,gender):
        self.name = name
        self.gender = gender
        
#outside the class-function that takes an object as a parameter

def greet(person):
    
    person.name = 'Adesh Srivastava'
    return person

p = Person('Aditya', 'Male')
print(id(p))
p1 = greet(p)
print(id(p1))

#here both The IDs will be the same because the classes, user-defined classes specifically, are mutable, so the changes made are stored at the same reference address. 
```
